```python
from datetime import datetime, timedelta
import pandas as pd
from gnews import GNews
import newspaper
from tqdm import tqdm

# Inisialisasi GNews
gnews = GNews(language="id", country="ID", max_results=1000000000000000000)

# Tentukan tanggal mulai dan tanggal akhir
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 5, 31)

# Fungsi untuk mendapatkan deskripsi lengkap menggunakan newspaper3k
def get_full_article_description(url):
    try:
        article = newspaper.Article(url)
        article.download()
        article.parse()
        return article.text  # Mengambil seluruh teks artikel
    except Exception as e:
        print(f"Error: {e}")
        return None

# Simpan hasil berita ke dalam list
news_list = []

# Loop untuk mengumpulkan berita dalam interval mingguan
current_start_date = start_date
while current_start_date < end_date:
    current_end_date = current_start_date + timedelta(days=7)
    if current_end_date > end_date:
        current_end_date = end_date

    # Debugging tanggal interval
    print(f"Mengambil artikel berita dari {current_start_date.date()} hingga {current_end_date.date()}")

    # Setel tanggal mulai dan tanggal akhir pada objek GNews untuk interval ini
    gnews.start_date = current_start_date
    gnews.end_date = current_end_date

    # Cari berita tentang inflasi di Gunungkidul
    query = "ekonomi gunung kidul"
    berita = gnews.get_news(query)

    # Debugging jumlah hasil yang ditemukan
    print(f"Ditemukan {len(berita)} artikel berita dari {current_start_date.date()} hingga {current_end_date.date()}")

    # Menampilkan progress scraping dengan tqdm
    for index, item in enumerate(tqdm(berita, desc=f"Mengambil artikel berita dari {current_start_date.date()} hingga {current_end_date.date()}")):
        print(f"Mengambil artikel berita {index + 1}/{len(berita)}: {item['title']}")
        print(f"URL: {item['url']}")
        deskripsi = get_full_article_description(item['url'])
        if not deskripsi:
            deskripsi = item['description']
        news_item = {
            'Judul': item['title'],
            'Deskripsi': deskripsi,
            'Tanggal Terbit': item['published date'],
            'URL': item['url'],
            'Penerbit': item['publisher']['title']
        }
        news_list.append(news_item)
        print(f"Telah terkumpul {len(news_list)} artikel berita.")

    # Perbarui tanggal mulai untuk interval berikutnya
    current_start_date = current_end_date + timedelta(days=1)

# Buat DataFrame dari list berita
df = pd.DataFrame(news_list)

# Buat nama file dari start_date, end_date, dan query
start_date_str = start_date.strftime("%Y-%m-%d")
end_date_str = end_date.strftime("%Y-%m-%d")
query_str = query.replace(" ", "_")
filename = f"{start_date_str}_{end_date_str}_{query_str}.xlsx"

# Simpan DataFrame ke file Excel
df.to_excel(filename, index=False)

print(f"Data berhasil disimpan ke dalam file '{filename}'")
```

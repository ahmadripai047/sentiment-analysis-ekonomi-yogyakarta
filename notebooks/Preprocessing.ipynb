```python
import pandas as pd
import re

# Membaca file CSV
data = pd.read_csv("combined_cleaned_data.csv")

# Menampilkan informasi tentang dataset
data.info()
```

```python
# Konversi kolom publish_date ke datetime
data['publish_date'] = pd.to_datetime(data['publish_date'], errors='coerce')

# Tentukan rentang waktu
start_date = pd.to_datetime("2020-01-01")
end_date = pd.to_datetime("2024-12-31")

# Tentukan keyword
keyword = ["ekonomi yogyakarta","inflasi yogyakarta","perdagangan yogyakarta","pariwisata yogyakarta"]

# Gabungkan keyword dengan regex
regex_pattern = '|'.join(keyword)

# Filter data berdasarkan rentang waktu dan keyword
filtered_data = data[
    (data['publish_date'] >= start_date) &
    (data['publish_date'] <= end_date) &
    (data['keyword'].str.contains(regex_pattern, case=False, na=False))
]

# Tampilkan data yang telah difilter
print(filtered_data)
```

```python
# Mengonversi kolom 'Tanggal Terbit' menjadi tipe datetime
df['publish_date'] = pd.to_datetime(df['publish_date'], format='%Y-%m-%d %H:%M:%S')

# Membuat kolom 'Month' untuk periode bulanan
df['Period'] = df['publish_date'].dt.to_period('M')

# Memisahkan bulan dan tahun menjadi kolom terpisah
df['Bulan'] = df['publish_date'].dt.month
df['Tahun'] = df['publish_date'].dt.year

# Menampilkan DataFrame
df.head()
```

```python
# Fungsi untuk menghapus nama kota dan penerbit
def remove_city_and_publisher(text):
    # Pola regex untuk mendeteksi nama kota dan penerbit di awal teks
    pattern = r'^\w+,\s+\w+\s'
    # Menghapus pola tersebut dari teks
    cleaned_text = re.sub(pattern, '', text)
    return cleaned_text

# Terapkan pada kolom 'Deskripsi'
df['content'] = df['content'].apply(remove_city_and_publisher)
```

```python
# Fungsi untuk membersihkan teks ulasan
def cleaningtext(text):
    text = re.sub(r'@[A-Za-z0-9]+', ' ', text)  # Menghapus @mentions
    text = re.sub(r'#[A-Za-z0-9]+', ' ', text)  # Menghapus hashtags
    text = re.sub(r"http\S+|www\S+|https\S+", ' ', text, flags=re.MULTILINE)  # Menghapus URL
    text = re.sub(r'[0-9]+', ' ', text)  # Menghapus angka
    text = re.sub(r'\n', ' ', text)  # Menghapus baris baru
    text = re.sub(r"[-()\"#/@;:<>{}'+=~|.!?,_]", " ", text)  # Menghapus karakter khusus
    text = re.sub(r'[^\x00-\x7F]+', ' ', text) # Menghapus karakter non-ASCII (untuk Unicode yang tidak diinginkan)
    text = re.sub(r'\s+', ' ', text) #Menghapus spasi ganda dan memastikannya hanya ada satu spasi
    text = text.strip()  # Menghapus spasi di awal dan akhir
    return text  # Pastikan return di dalam fungsi

# Menerapkan preprocessing pada kolom 'Clean_Text'
df['Clean_Text'] = df['content'].apply(cleaningtext)

# Menampilkan hasil beberapa kolom yang sudah dibersihkan
print(df[['Clean_Text']])
```

```python
# Fungsi untuk merubah teks menjadi lowercase
def casefoldingText(text):
    text = text.lower()
    return text

# Menerapkan fungsi casefoldingText pada kolom '3/Lebih' dari dataframe df
df['Clean_Text'] = df['Clean_Text'].apply(casefoldingText)

# Menampilkan hasil perubahan lowercase
print(df[['Clean_Text']])
```

```python
# Fungsi untuk menghapus kata-kata yang terlalu pendek
def hapus_kata_pendek(teks, batas_panjang=2):
    # Tokenisasi teks menjadi kata-kata
    kata_kata = teks.split()
    # Menghapus kata-kata yang panjangnya kurang dari batas_panjang
    kata_kata_bersih = [kata for kata in kata_kata if len(kata) >= batas_panjang]
    return ' '.join(kata_kata_bersih)

# Fungsi untuk menghapus kalimat yang tidak relevan
def hapus_kalimat_tidak_relevan(teks):
    # Hapus kalimat yang hanya berisi karakter non-alfabet atau terlalu pendek
    if not teks or len(teks.split()) < 2:
        return ''
    return teks

# Menerapkan fungsi ke dalam DataFrame
df['Clean_Text'] = df['Clean_Text'].apply(hapus_kata_pendek)  # Menghapus kata-kata pendek
df['Clean_Text'] = df['Clean_Text'].apply(hapus_kalimat_tidak_relevan)  # Menghapus kalimat yang tidak relevan

# Menampilkan hasil
print(df[['Clean_Text']])
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Fungsi untuk membaca stopwords dari file
def load_stopwords(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        stopwords = file.read().splitlines()  # Membaca setiap baris sebagai stopword
    return set(stopwords)

# Path ke file stopwords
stopwords_path = r'C:\Users\asus\Documents\MUH RIFAI\Tugas Akhir\TA1\stopwords.indo.txt'

# Memuat stopwords dari file
stopwords_indonesia = load_stopwords(stopwords_path)

# Menambahkan kata-kata tambahan ke dalam stopwords
additional_stopwords = {'dengan', 'untuk','jadi'}
stopwords_indonesia.update(additional_stopwords)

# Fungsi untuk menghapus stopwords dari teks
def hapus_stopwords(teks, stopwords):
    # Tokenisasi teks
    kata_kata = word_tokenize(teks)
    # Menghapus stopwords
    kata_kata_bersih = [kata for kata in kata_kata if kata not in stopwords]
    # Gabungkan token menjadi string kembali
    return ' '.join(kata_kata_bersih)

# Terapkan penghapusan stopwords pada kolom 'Clean_Text' di dataframe
df['Clean_Text'] = df['Clean_Text'].apply(lambda x: hapus_stopwords(x, stopwords_indonesia))

# Menampilkan hasil setelah penghapusan stopwords
print(df[['Clean_Text']])
```

```python
import re

# Path ke file slangwords.txt
kamus_slang_path = r'C:\Users\asus\Documents\MUH RIFAI\Tugas Akhir\TA1\slangwords.txt'

# Kamus Typo yang sudah Anda buat
kamus_typo = {
    'yg': 'yang', 'rp':'rupiah','ga': 'tidak', 'gak': 'tidak', 'gk': 'tidak', 'tdk': 'tidak',
    'dlm': 'dalam', 'sm': 'sama', 'klo': 'kalau', 'aja': 'saja', 'dr': 'dari',
    'trs': 'terus', 'tp': 'tapi', 'udh': 'sudah', 'jg': 'juga', 'blm': 'belum',
    'sy': 'saya', 'aku': 'saya', 'km': 'kamu', 'lg': 'lagi', 'bnr': 'benar',
    'bs': 'bisa', 'hrs': 'harus', 'msh': 'masih', 'org': 'orang', 'skrg': 'sekarang',
    'ny': 'nya', 'knp': 'kenapa', 'gmna': 'bagaimana', 'pdhl': 'padahal',
    'smg': 'semoga', 'krn': 'karena', 'mlh': 'malah', 'ntr': 'nanti', 'sdh': 'sudah',
    'bgmn': 'bagaimana', 'tlg': 'tolong', 'banget': 'sangat', 'mau': 'ingin',
    'kasih': 'beri', 'seler': 'penjual', 'makasih': 'terima kasih',
    'btw': 'ngomong-ngomong', 'brg': 'barang', 'jd': 'jadi', 'bgt': 'sangat',
    'dmn': 'di mana', 'nggak': 'tidak', 'ajaib': 'menakjubkan', 'amanah': 'dapat dipercaya',
    'pokoknya': 'intinya', 'keren': 'luar biasa', 'mantap': 'hebat', 'ajib': 'menakjubkan',
    'kece': 'menarik', 'cuan': 'keuntungan', 'ribet': 'rumit', 'copot': 'terlepas',
    'lapor': 'melapor', 'cepet': 'cepat', 'capek': 'lelah', 'sblm': 'sebelum',
    'trm': 'terima', 'gampang': 'mudah', 'lumayan': 'cukup', 'ane': 'saya',
    'agan': 'bapak/ibu', 'gt': 'gitu', 'dng': 'dengan', 'trimakasih': 'terima kasih',
    'gmn': 'bagaimana', 'udah': 'sudah', 'jir': 'anjir', 'janggan': 'jangan',
    'ama': 'sama', 'bosok': 'busuk', 'hadeuhh': 'aduh','pt':'perusahaan','to':'untuk','with':'dengan'
}

# Membaca file slangwords.txt dan mengonversinya menjadi kamus
def load_slang(kamus_slang_path):
    kamus_slang = {}
    with open(kamus_slang_path, 'r', encoding='utf-8') as file:
        for line in file:
            if '=>' in line:
                slang, baku = line.strip().split('=>')
                kamus_slang[slang.strip()] = baku.strip()
    return kamus_slang

# Load kamus slang
kamus_slang = load_slang(kamus_slang_path)

# Fungsi untuk mengganti slang dengan bentuk baku
def ubah_slang(teks, kamus_slang):
    words = teks.split()
    processed_words = [kamus_slang.get(word, word) for word in words]
    return ' '.join(processed_words)

# Fungsi untuk memperbaiki typo
def perbaiki_typo(teks, kamus_typo):
    words = teks.split()
    processed_words = [kamus_typo.get(word, word) for word in words]
    return ' '.join(processed_words)

# Gabungkan keduanya: slang dan typo correction
def normalisasi_teks(teks, kamus_slang, kamus_typo):
    teks = ubah_slang(teks, kamus_slang)  # Ganti slang
    teks = perbaiki_typo(teks, kamus_typo)  # Perbaiki typo
    return teks

# Terapkan normalisasi pada kolom 'Clean_Text' di dataframe
df['Clean_Text'] = df['Clean_Text'].apply(lambda x: normalisasi_teks(x, kamus_slang, kamus_typo))

# Menampilkan hasil setelah penghapusan slang dan typo correction
print(df[['Clean_Text']])
```
